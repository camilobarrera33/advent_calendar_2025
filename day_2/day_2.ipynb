```python
# Read input data from file
with open("input.txt", "r") as f:
    content = f.read()

# Split the content into each individual range
# Each range is separated by a comma
ranges = content.split(",")

# Create a list to hold all the numbers that meet the criteria
result_list = []

# Loop through all ranges
for i in ranges:
    # Split the string to determine starting and ending point of the range
    strings = i.split("-")
    min_ = int(strings[0])
    max_ = int(strings[-1])
    # Loop through all numbers in the range
    for j in range(min_, max_+1):
        j_string = str(j)
        # The condition is only met if the string length is even
        if len(j_string) % 2 == 0:
            # Find the midpoint of the string
            mid = len(j_string) // 2
            # Compare each half of the string
            # If they are the same, add the number to the result list
            if j_string[:mid] == j_string[mid:]:
                result_list.append(j)


# Sum all the number in the `result_list` to obtain the final answer to the problem
final_result = sum(result_list)
print(f"Final result: {final_result}")

```
