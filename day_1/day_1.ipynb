# Part One

```python
def check_bounds(num):
    if num < 0:
        factor = -(num//100)
        return num + (100 * factor)
    elif num > 99:
        factor = num//100
        return num - (100 * factor)
    else:
        return num

position = 50
zero_count = 0

with open("input.txt", "r") as f:
    lines = f.readlines()

for line in lines:
    change = int(line[1:])
    if line[0] == 'R':
        position += change
        position = check_bounds(position)
    else:
        position -= change
        position = check_bounds(position)
    if position == 0:
        zero_count += 1

print(position)
print(zero_count)
```

# Part Two

```python
def check_bounds(num):
    if num < 0:
        factor = -(num//100)
        return num + (100 * factor), factor
    elif num > 99:
        factor = num//100
        return num - (100 * factor), factor
    else:
        return num, 0

position = 50
zero_count = 0
factor_count = 0

with open("input_2.txt", "r") as f:
    lines = f.readlines()

for line in lines:
    change = int(line[1:])
    if line[0] == 'R':
        position += change
        position, factor = check_bounds(position)
        factor_count += factor
    else:
        position -= change
        position, factor = check_bounds(position)
        factor_count += factor
    if position == 0:
        zero_count += 1

print(position)
print(zero_count)
print(factor_count)
print(zero_count + factor_count)
```
